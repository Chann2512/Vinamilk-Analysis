# **02. Tiền xử lý và Chuẩn hóa Văn bản (Text Preprocessing)**

## **Mục tiêu:**
- Áp dụng các kỹ thuật làm sạch văn bản tiếng Việt:
  - Chuẩn hóa Unicode NFKC và loại bỏ chữ fancy.
  - Xóa URL, các biểu tượng cảm xúc (Emoji).
  - Xóa các ký tự đặc biệt không mang ngữ nghĩa.
  - Thu gọn ký tự lặp liên tiếp (ví dụ: `ngonnn` -> `ngon`).
  - Dịch các từ viết tắt và tiếng lóng mua sắm (slang).
  - Loại bỏ các từ dừng tiếng Việt (Stopwords).

```python
import os
import sys
import pandas as pd

sys.path.append(os.path.abspath('../src'))
import utils
from preprocessing import clean_text
```

### **1. Nạp tập dữ liệu bình luận chữ**

```python
df = pd.read_csv("../data/processed/text_reviews.csv", encoding="utf-8-sig")
print(f"Đang xử lý {len(df):,} dòng bình luận có chữ.")
```

### **2. Chạy thử nghiệm hàm làm sạch**
Kiểm tra tính chính xác của hàm `clean_text` trên các bình luận mẫu đặc thù của Shopee.

```python
sample_texts = [
    "Sữa ngonnn lắm nha mọi người, đóng gói rất oke, giao hàng nhanh cực kì!!! ❤️❤️❤️",
    "Sản phẩm k giống mô tả, shop tư vấn quá chán, ko mua lại nữa. Link: https://shopee.vn/test",
    "Hàng rất tốt oki oki oki, date xa, sẽ tiếp tục ủng hộ shop."
]

print("--- KẾT QUẢ TIỀN XỬ LÝ MẪU ---")
for text in sample_texts:
    cleaned = clean_text(text)
    print(f"🔹 Gốc: {text}")
    print(f"👉 Sạch: {cleaned}\n")
```

### **3. Chạy làm sạch trên toàn bộ dữ liệu**
Thực hiện biến đổi cột `comment` và lưu vào cột mới `cleaned_comment`.

```python
print("🚀 Bắt đầu làm sạch dữ liệu lớn (quá trình này có thể mất ít phút)...")
df["cleaned_comment"] = df["comment"].astype(str).apply(lambda x: clean_text(x, remove_sw=True))
print("✅ Hoàn thành tiền xử lý văn bản!")

# Hiển thị ngẫu nhiên kết quả để đối chiếu
display(df[["comment", "cleaned_comment"]].sample(5))
```

### **4. Lưu trữ dữ liệu sạch**
Lưu DataFrame đã làm sạch vào file mới.

```python
df.to_csv("../data/processed/clean_reviews.csv", index=False, encoding="utf-8-sig")
print("✅ Lưu dữ liệu sạch thành công tại ../data/processed/clean_reviews.csv")
```
