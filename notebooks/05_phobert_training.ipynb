### **1. Giải thích về quy trình cân bằng dữ liệu**
Do lớp Tích cực (2) chiếm đa số tuyệt đối, việc huấn luyện trực tiếp sẽ làm mô hình bị thiên lệch.
```python
data_path = "../data/processed/train_dataset.csv"
output_model_dir = "../models/phobert_sentiment"

print("🚀 Bắt đầu tiến trình huấn luyện và tinh chỉnh PhoBERT...")
train_model(
    data_path=data_path,
    output_dir=output_model_dir,
    model_name="vinai/phobert-base",
    epochs=3,
    batch_size=4
)
print("🎉 Quá trình huấn luyện đã kết thúc thành công!")
```
